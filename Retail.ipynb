```python
# Step 1: Define the ASCII sum function
def ascii_sum(code):
    return sum(ord(char) for char in str(code))

# Step 2: Apply it to the 'StockCode' column
df['StockCode_ASCII'] = df['StockCode'].astype(str).apply(ascii_sum)
```

```python
df = df.drop('StockCode', axis = 1)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Step 1: Clean or fill NaNs
df['Description'] = df['Description'].fillna('')

# Step 2: TF-IDF Vectorizer
tfidf = TfidfVectorizer(max_features=100)  # You can increase or reduce this
X_tfidf = tfidf.fit_transform(df['Description'])

# Step 3: Merge into your DataFrame
tfidf_df = pd.DataFrame(X_tfidf.toarray(), columns=tfidf.get_feature_names_out())
df = pd.concat([df.reset_index(drop=True), tfidf_df.reset_index(drop=True)], axis=1)

```
